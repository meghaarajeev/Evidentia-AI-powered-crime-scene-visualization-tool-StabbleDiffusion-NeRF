```python
import fitz  # PyMuPDF

def extract_text_from_pdf(pdf_path):
    text = ""
    doc = fitz.open(pdf_path)  # Open the PDF
    for page in doc:
        text += page.get_text("text") + "\n"
    return text

# Change 'fir_short.pdf' and 'fir_long.pdf' to your uploaded filenames
pdf_short = "FIR short.pdf"
pdf_long = "FIR NEW LONG 2.pdf"

short_fir_text = extract_text_from_pdf(pdf_short)
long_fir_text = extract_text_from_pdf(pdf_long)

print("✅ Short FIR Extracted:\n", short_fir_text)
print("\n✅ Long FIR Extracted:\n", long_fir_text)
```

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

# Load T5 model and tokenizer
model_name = "t5-small"  # Can be changed to "t5-base" or "facebook/bart-large-cnn"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)

def generate_text(fir_text, mode="expand"):
    """
    Expands short FIR text into 2-3 detailed paragraphs or condenses long FIR into key details.

    :param fir_text: Extracted FIR text.
    :param mode: 'expand' for detailed text, 'summarize' for key details.
    :return: Generated text.
    """
    if not fir_text.strip():
        return "No valid FIR text provided."

    if mode == "expand":
        task_prefix = "Generate a detailed incident report: "  # Stronger instruction
        max_length = 500  # Increase to allow longer output
        min_length = 200  # Ensure multiple paragraphs
        temperature = 0.7  # Increases creativity in output
    elif mode == "summarize":
        task_prefix = "Summarize the key details of this FIR: "
        max_length = 100
        min_length = 50
        temperature = 0.3  # Keeps summarization concise
    else:
        raise ValueError("Mode must be either 'expand' or 'summarize'.")

    # Tokenize and generate text
    inputs = tokenizer(task_prefix + fir_text, return_tensors="pt", max_length=512, truncation=True)
    output_ids = model.generate(
        inputs.input_ids,
        max_length=max_length,
        min_length=min_length,
        num_beams=7,  # More beams for better quality
        length_penalty=2.5,  # Encourages verbosity
        early_stopping=True,
        temperature=temperature,  # Controls randomness
        repetition_penalty=1.2  # Avoids repetition
    )

    return tokenizer.decode(output_ids[0], skip_special_tokens=True)


# Example FIR reports (Replace with extracted FIR text)
short_fir_text = """
On 23rd February 2025, at 9:30 PM, my mobile phone was snatched by two men on a bike near MG
Road.
"""
long_fir_text = """
On 23rd February 2025, at approximately 9:30 PM, I was walking home from work near MG Road,
when two unidentified individuals on a black Pulsar motorcycle approached me from behind. The
pillion rider swiftly snatched my mobile phone from my hand and the vehicle sped away towards the
main highway.

The street was dimly lit, and I could not see their faces clearly. However, the rider was wearing a
black helmet, and the passenger had a red jacket. The incident happened very quickly, and I was
unable to react in time. A passerby mentioned seeing a similar bike loitering around the area earlier
in the evening.

I immediately informed the police by calling emergency services and provided details of the event.
My phone contained important personal and professional data. I request the authorities to check
CCTV footage from nearby shops and street cameras to identify the suspects.
"""

# Generate detailed and summarized FIR
detailed_fir = generate_text(short_fir_text, mode="expand")  # Fixed: Now calling the function properly
summarized_fir = generate_text(long_fir_text, mode="summarize")

# Print results
print("\n🔹 Detailed FIR :\n", detailed_fir)
print("\n🔹 Summarized FIR :\n", summarized_fir)
```

```python
%%writefile backend.py

from flask import Flask, request, jsonify
import fitz  # PyMuPDF
from transformers import T5Tokenizer, T5ForConditionalGeneration

# Load T5 model and tokenizer
model_name = "t5-small"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)

app = Flask(__name__)

# Function to extract text from PDF
def extract_text_from_pdf(pdf_path):
    text = ""
    doc = fitz.open(pdf_path)
    for page in doc:
        text += page.get_text("text") + "\n"
    return text

# Function to generate expanded or summarized FIR text
def generate_text(fir_text, mode="expand"):
    if not fir_text.strip():
        return "No valid FIR text provided."

    if mode == "expand":
        task_prefix = "Generate a detailed incident report: "
        max_length = 500
        min_length = 200
        temperature = 0.7
    elif mode == "summarize":
        task_prefix = "Summarize the key details of this FIR: "
        max_length = 100
        min_length = 50
        temperature = 0.3
    else:
        return "Invalid mode"

    inputs = tokenizer(task_prefix + fir_text, return_tensors="pt", max_length=512, truncation=True)
    output_ids = model.generate(
        inputs.input_ids,
        max_length=max_length,
        min_length=min_length,
        num_beams=7,
        length_penalty=2.5,
        early_stopping=True,
        temperature=temperature,
        repetition_penalty=1.2
    )

    return tokenizer.decode(output_ids[0], skip_special_tokens=True)

# API Endpoint to process FIR PDFs
@app.route("/process_fir", methods=["POST"])
def process_fir():
    if "file" not in request.files:
        return jsonify({"error": "No file uploaded"}), 400

    file = request.files["file"]
    if file.filename == "":
        return jsonify({"error": "Empty file"}), 400

    file_path = "./temp_fir.pdf"
    file.save(file_path)

    extracted_text = extract_text_from_pdf(file_path)
    detailed_fir = generate_text(extracted_text, mode="expand")
    summarized_fir = generate_text(extracted_text, mode="summarize")

    return jsonify({
        "extracted_text": extracted_text,
        "detailed_fir": detailed_fir,
        "summarized_fir": summarized_fir
    })

if __name__ == "__main__":
    app.run(debug=True)
```

```python
from google.colab import files
import fitz  # PyMuPDF
from transformers import T5Tokenizer, T5ForConditionalGeneration
from datasets import load_metric

# Upload PDF files
uploaded = files.upload()

# Function to extract text from PDFs
def extract_text_from_pdf(pdf_path):
    text = ""
    doc = fitz.open(pdf_path)
    for page in doc:
        text += page.get_text("text") + "\n"
    return text

# Change these filenames as per your uploaded files
pdf_short = "FIR short.pdf"
pdf_long = "FIR NEW LONG 2.pdf"

# Extract FIR texts
short_fir_text = extract_text_from_pdf(pdf_short)
long_fir_text = extract_text_from_pdf(pdf_long)

print("✅ Short FIR Extracted:\n", short_fir_text)
print("\n✅ Long FIR Extracted:\n", long_fir_text)

# Load T5 model and tokenizer
model_name = "t5-small"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)

# Text generation function
def generate_text(fir_text, mode="expand"):
    if not fir_text.strip():
        return "No valid FIR text provided."

    if mode == "expand":
        task_prefix = "Generate a detailed incident report: "
        max_length = 500
        min_length = 200
        temperature = 0.7
    elif mode == "summarize":
        task_prefix = "Summarize the key details of this FIR: "
        max_length = 100
        min_length = 50
        temperature = 0.3
    else:
        raise ValueError("Mode must be either 'expand' or 'summarize'.")

    inputs = tokenizer(task_prefix + fir_text, return_tensors="pt", max_length=512, truncation=True)
    output_ids = model.generate(
        inputs.input_ids,
        max_length=max_length,
        min_length=min_length,
        num_beams=7,
        length_penalty=2.5,
        early_stopping=True,
        temperature=temperature,
        repetition_penalty=1.2
    )

    return tokenizer.decode(output_ids[0], skip_special_tokens=True)

# Generate FIR reports
detailed_fir = generate_text(short_fir_text, mode="expand")
summarized_fir = generate_text(long_fir_text, mode="summarize")

print("\n🔹 Detailed FIR :\n", detailed_fir)
print("\n🔹 Summarized FIR :\n", summarized_fir)

### PERFORMANCE EVALUATION ###
rouge = load_metric("rouge")
bleu = load_metric("bleu")

def evaluate_model(predictions, references):
    # Compute ROUGE scores
    rouge_scores = rouge.compute(predictions=predictions, references=references)

    # Compute BLEU score
    bleu_scores = bleu.compute(predictions=[pred.split() for pred in predictions],
                               references=[[ref.split()] for ref in references])

    return {
        "ROUGE-1": rouge_scores["rouge1"].mid.fmeasure,
        "ROUGE-2": rouge_scores["rouge2"].mid.fmeasure,
        "ROUGE-L": rouge_scores["rougeL"].mid.fmeasure,
        "BLEU": bleu_scores["bleu"]
    }

# Evaluate detailed FIR (comparison with long FIR)
detailed_fir_scores = evaluate_model([detailed_fir], [long_fir_text])
print("\n📊 Performance of Expanded FIR:\n", detailed_fir_scores)

# Evaluate summarized FIR (comparison with short FIR)
summarized_fir_scores = evaluate_model([summarized_fir], [short_fir_text])
print("\n📊 Performance of Summarized FIR:\n", summarized_fir_scores)
```
